## Problem statement

>
> Six letters are chosen at random from the ones of the word PROPORTION. Determine the probability
that it is possible to form the following words from these letters:
>
> (a) OPTION;
>
> (b) PORT;
>
> (c) RIOT.


## But how to solve it by hand?
First, let us count how many times each letter appears:

| Letter | P | R | O | T | I | N |
|--------|---|---|---|---|---|---|
| Counts | 2 | 2 | 3 | 1 | 1 | 1 |

In total, there are $C_10^6 = \frac{10!}{6! \cdot 4!} = 210$ ways to choose 6 letters out of 10 given.

### Case a: OPTION
We need to choose 2 O's out of 3 available. It can be done in $C_3^2 = 3$ ways. Further, we need to choose one of two P's available. It can be accomplished in $C_2^1 = 2$ ways. All remaining letters appear just once. Further, as any selection of 2 O's is compatible with any selection of P, in total we have $C_3^2 \cdot C_2^1 = 6$ favorable outcomes.

Thus, $P(OPTION) = \frac{6}{210} = \frac{1}{35}$

### Case b: PORT
Let X denote letter I or letter N. All favorable outcomes follow one of the following patterns:

| Pattern | Number of outcomes |
|---------|--------------------|
| PORTOO  | $C_2^1 \cdot C_2^1 = 4$ (choosing P & R) |
| PORTOR  | $C_2^1 \cdot C_3^2 = 6$ (choosing P & 2 O's) |
| PORTOP  | $C_2^1 \cdot C_3^2 = 6$ (choosing R & 2 O's) |
| PORTOX  | $C_2^1 \cdot C_2^1 \cdot C_3^2 \cdot C_2^1 = 24$ (chosing P, R, 2 O's and X) |
| PORTPR  | $C_3^1 = 3$ (choosing O) |
| PORTPX  | $C_2^1 \cdot C_3^1 \cdot C_2^1 = 12$ (choosing R, O, X) |
| PORTRX  | $C_2^1 \cdot C_3^1 \cdot C_2^1 = 12$ (choosing P, O, X) |
| PORTIN  | $C_2^1 \cdot C_3^1 \cdot C_2^1 = 12$ (choosing P, O, R) |

*Note:* for shortness terms which are identically one (e.g. $C_1^1$ and $C_3^3$) are omitted.

As we could see from the table above, there are $4 + 2 \cdot 6 + 24 + 3 + 3 \cdot 12 = 79$ favorable outcomes. Thus, $P(PORT) = \frac{79}{210}$.

### Case c: RIOT
| Pattern | Number of outcomes |
|---------|--------------------|
| RIOTOO  | $C_2^1 = 2$ (choosing R) |
| RIOTOR  | $C_3^1 = 3$ (choosing 2 O's) |
| RIOTOP  | $C_2^1 \cdot C_3^2 \cdot C_2^1 = 12$ (choosing P, 2 O's, R) |
| RIOTPP  | $C_2^1 \cdot C_3^1 = 6$ (choosing R & O) |
| RIOTPR  | $C_3^1 \cdot C_2^1 = 6$ (choosing O & P) |
| RIOTON  | $C_2^1 \cdot C_3^2 = 6$ (choosing R & 2 O's) |
| RIOTRN  | $C_3^1 = 3$ (choosing O) |
| RIOTPN  | $C_1 \cdot C_3^1 \cdot C_2^1 = 12$ (choosing R, O, P) |

Thus, in total there are $2 + 2 \cdot 3 + 2 \cdot 12 + 3 \cdot 6 = 50$ favorable outcomes. $P(RIOT) = \frac{50}{210} = \frac{5}{21}$.


## Necessary imports and constants

```python
from itertools import combinations
from collections import Counter
from typing import List

BASE_WORD = "PROPORTION"
```

## Implementing solver
1. How to check if a target word can be constructed from a given collection of chosen letters?
2. How to produce all possible letters selection?
3. How to find a probability?

Let us try to solve the problem step-by-step.For the sake of simplicity, we will assume that all letters are capitalized.

### Step 1: checking if a target word can be obtained using chosen letters
<font color="red">TODO:</font> complete the function that checks if we can construct a target word using chosen letters.

*Hint:* `Counter` $A$ is less than or equal to `Counter` $B$, if for any element $a$ of $A$ $counts_A{a} \leq counts_B{a}$. Where $counts_X(a)$ denotes the counts of element $a$ in `Counter` $X$.

```python
def can_be_written(target_word: str, chosen_letters: List[str]) -> bool:
  """
  target_word - the word that we aim to construct
  chosen_letters - list of the letters chosen

  Returns: True, if we can write the target word using chosen_letters and False otherwise
  """
  assert all([len(c) == 1 for c in chosen_letters]), "Chosen letters must be a list of characters or a string"

  target_counts = Counter(target_word)
  chosen_letters_counts = Counter(chosen_letters)

  return target_counts <= chosen_letters_counts
```

```python
assert can_be_written("RIOT", ["R", "I", "L", "O", "T"]), "Yes, word RIOT can be written with the use of letters RILOT"
assert not can_be_written("RIOT", ["P", "I", "L", "O", "T"]), "No, word RIOT can't be written with the use of letters PILOT"
```

## Step 2: counting outcomes
<font color="red">TODO:</font> complete the function that computes number of favorable outcomes and total number of outcomes using `can_be_written` function.

*Hint:* check [itertools.combinations documentation](https://docs.python.org/3/library/itertools.html#itertools.combinations).

```python
def count_outcomes(base_word: str, target_word: str, num_letters: int) -> tuple[int, int]:
  """
  base_word - the word from which we select letters
  target_word - the word that we want to write using letters from the base_word
  num_letters - number of the letters we take from the base word

  Returns: number of favorable outcomes and total number of outcomes
  """
  favorable_outcomes, total_outcomes = 0, 0

  for chosen_letters in combinations(base_word, num_letters):
    total_outcomes += 1
    if can_be_written(target_word, chosen_letters):
      favorable_outcomes += 1

  return favorable_outcomes, total_outcomes
```

```python
favorable, total = count_outcomes(BASE_WORD, "RIOT", 6)
assert favorable == 50, "Incorrect number of favorable outcomes"
assert total == 210, "Incorrect number of total outcomes"
```

## Step 3: calculating the probability
<font color="red">TODO:</font> use `count_outcomes` and classical definition of probability to obtain answers for all three problems.

```python
def solve_problem(base_word: str, target_word: str, num_letters: int) -> None:
  """
  base_word - the word from which we select letters
  target_word - the word that we want to write using letters from the base_word
  num_letters - number of the letters we take from the base word

  Returns: nothing. Prints an answer to a console in a format "<probability> (<number of favorable outcomes>/<total number of outcomes>)"
  """
  favorable, total = count_outcomes(base_word, target_word, num_letters)
  print(f"p = {favorable/total: .3f} ({favorable}/{total})")
```

```python
solve_problem(BASE_WORD, "OPTION", 6)
```

```python
solve_problem(BASE_WORD, "PORT", 6)
```

```python
solve_problem(BASE_WORD, "RIOT", 6)
```
